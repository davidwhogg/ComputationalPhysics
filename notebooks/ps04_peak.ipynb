```python
import numpy as np
import pickle
import pylab as plt
```

```python
xs = np.arange(1.5, 20., 1.0)
ys = np.ones_like(xs)
ys += np.exp(-0.5 * (xs - 13.65) ** 2) # magic: the number I always use in class
```

```python
rng = np.random.default_rng(17) # the most random number
ys += 0.03 * rng.normal(size=xs.shape)
```

```python
plt.scatter(xs, ys, color="k")
```

```python
with open("ps04_peak.pkl", "wb") as file:
    pickle.dump((xs, ys), file)
```
