```python
import numpy as np
```

```python
xs = np.array([1., ] * 30_000_000 + [1.e20, ] * 50 + [-1.e20, ] * 50)
print(xs.shape)
```

```python
print(xs)
```

```python
a = np.sum(xs)
print(a)
```

```python
b = np.sum(xs[::-1])
print(b)
```

```python
rng = np.random.default_rng(17)
rng.shuffle(xs)
print(xs)
```

```python
c = np.sum(xs)
print(c)
```

```python
d = np.sum(np.sort(xs))
```

```python
print(a - b, a - c, b - c, a - d)
```
