```python
import csv
import numpy as np
```

```python
xs = np.array([1., ] * 30_000_000 + [1.e20, ] * 50 + [-1.e20, ] * 50)
print(xs.shape)
```

```python
print(xs)
```

```python
a = np.sum(xs)
print(a)
```

```python
b = np.sum(xs[::-1])
print(b)
```

```python
rng = np.random.default_rng(17)
rng.shuffle(xs)
print(xs)
```

```python
# Open the file in write mode ('w')
with open('ps02.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerows(list(xs[:,None]))
```
