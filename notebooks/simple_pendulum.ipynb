```python
import numpy as np
import pylab as plt
```

```python
def leapfrog_start(r, v, h):
    """
    ## notes:
    - convention: move position is 0.5 * h into the future
    """
    return r + v * 0.5 * h, v + 0.

def leapfrog_step(r, v, t, a, h):
    v_next = v + a(r, t) * h
    return r + v_next * h, v_next

def leapfrog(r, v, a, h, n):
    """
    ## bugs:
    - custom termination criteria because we have no principles
    
    ## notes:
    - returns two sets of times, one for the positions and one for the velocities
    """
    d = len(np.atleast_1d(r))
    assert len(np.atleast_1d(v)) == d
    rs, vs, ts = np.zeros((n, d)) + np.nan, np.zeros((n, d)) + np.nan, h * np.arange(n)
    rr, vv = leapfrog_start(r, v, h)
    rs[0], vs[0] = rr, vv
    for i in range(1, n):
        rr, vv = leapfrog_step(rr, vv, ts[i], a, h)
        if rr > np.pi:
            break
        if vv < 0.:
            break
        rs[i], vs[i] = rr, vv
    return rs, ts + 0.5 * h, vs, ts
```

```python
# setting sqrt(g/l) = 1

def domega_dt(theta, t):
    return -np.sin(theta)
```

```python
kwargs = {"linestyle": "-", "color": "k", "lw": 0.5}
theta_0 = 0.
for omega_0 in np.arange(0.2, 4.1, 0.2):
    thetas, tts, omegas, tos = leapfrog(theta_0, omega_0, domega_dt, 0.01, 800)
    thetas = np.interp(tos, tts, thetas[:, 0])
    plt.plot(thetas, omegas, **kwargs)
    plt.plot(thetas + 2. * np.pi, omegas, **kwargs)
    plt.plot(thetas - 2. * np.pi, omegas, **kwargs)
    plt.plot(-thetas, omegas, **kwargs)
    plt.plot(-thetas + 2. * np.pi, omegas, **kwargs)
    plt.plot(-thetas - 2. * np.pi, omegas, **kwargs)
    plt.plot(thetas, -omegas, **kwargs)
    plt.plot(thetas + 2. * np.pi, -omegas, **kwargs)
    plt.plot(thetas - 2. * np.pi, -omegas, **kwargs)
    plt.plot(-thetas, -omegas, **kwargs)
    plt.plot(-thetas + 2. * np.pi, -omegas, **kwargs)
    plt.plot(-thetas - 2. * np.pi, -omegas, **kwargs)
plt.xlim(-2. * np.pi, 2. * np.pi)
plt.ylim(-2.5, 2.5)
plt.xlabel(r"angle $\theta$ [rad]")
plt.ylabel(r"angular velocity $\omega$ [rad s$^{-1}$]")
```
